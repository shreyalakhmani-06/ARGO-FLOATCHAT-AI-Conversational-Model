```python
import os
import pandas as pd

# Folder paths
raw_folder = r"C:\Users\manas\Desktop\MVP1\csv_data\profiles"              # folder with 15 raw CSVs
cleaned_folder = "cleaned_profile_data"  # output folder

os.makedirs(cleaned_folder, exist_ok=True)

def clean_binary_strings(value):
    """Convert b'XXX' → XXX, leave other values unchanged."""
    if isinstance(value, str) and value.startswith("b'") and value.endswith("'"):
        return value[2:-1]  # remove b' and trailing '
    return value

# Process all CSVs
for file in os.listdir(raw_folder):
    if file.endswith(".csv"):
        print(f"Cleaning {file} ...")

        # Read CSV
        df = pd.read_csv(os.path.join(raw_folder, file))

        # Apply cleaning to all string-like values
        df = df.applymap(clean_binary_strings)

        # Save to cleaned folder
        cleaned_path = os.path.join(cleaned_folder, file)
        df.to_csv(cleaned_path, index=False)

print("✅ Cleaning complete! Files saved in 'cleaned_profile_data/'")
```
