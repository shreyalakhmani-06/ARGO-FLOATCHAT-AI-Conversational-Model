```python
import pandas as pd

# Load your merged floats file
file_path = "merged_floats.csv"   # change path if needed
df = pd.read_csv(file_path)

# Function to clean binary-like array strings into normal text
def clean_binary_array(value):
    if isinstance(value, str):
        value = (
            value.replace("[", "")
                 .replace("]", "")
                 .replace("b'", "")
                 .replace("'", "")
                 .replace("--", "")
                 .strip()
        )
        return value.replace(" ", "")
    return value

# Apply cleaning to all object columns
df_clean = df.copy()
for col in df_clean.columns:
    if df_clean[col].dtype == object:
        df_clean[col] = df_clean[col].astype(str).apply(clean_binary_array)

# Save to a new CSV
output_path = "merged_floats_clean.csv"
df_clean.to_csv(output_path, index=False)

print(f"Cleaned CSV saved to: {output_path}")
```
